```python
#first simple plot (doesnt really give us anything valuable since its just iterating through the rows from top to bottom and diplaying their temps.)

import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/20210703_greenhouse_data.csv', sep=';', decimal=',')  #make sure to read dataset to see seperation and decimals charachters.

plt.figure(figsize=(12,5))
plt.plot(df["greenhous_temperature_celsius"])  #draw point markers and connect them with lines
plt.title("Temperature Over Time")
plt.xlabel("Index") #number of the row
plt.ylabel("Temperature (°C)")
plt.show() # Its primary purpose is to make the generated plots visible to the user.
```

```python
#second plot: Histogram of temperature

plt.figure()
plt.hist(df["greenhous_temperature_celsius"], bins=30)  #hist is used to generate histograms. divides the range of these values into a series of intervals called "bins"
plt.title("Temperature distribution")
plt.xlabel("Temperature (°C)")
plt.ylabel("Count")  #Number of rows (data points) that fall into each temperature interval (bin)
plt.show()
```

```python
# Third plot: Scatter of temperature vs humidity

plt.figure()
plt.scatter(df["greenhous_temperature_celsius"], df["greenhouse_humidity_percentage"])
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.show()
```
